```python
# get kingkong data
kingkong = data[data['Name'] == 'King Kong']
kingkong
```

```python
# drop duplicates
data = data.drop_duplicates()
data.shape
```

```python
# remove , from votes feature
data['Votes'] = data['Votes'].str.replace(',', '')
data.head()
```

```python
data["Rate"] = pd.to_numeric(data["Rate"], errors='coerce')
data["Votes"] = pd.to_numeric(data["Votes"], errors='coerce')
data.info()
```

```python
data.isnull().sum()
```

```python
# replace Rate, Vote, Duration null with 0
data['Rate'] = data['Rate'].fillna(0)
data['Votes'] = data['Votes'].fillna(0)
data['Duration'] = data['Duration'].fillna(0)
data.isnull().sum()
```

```python
# in categorical features, replace with "None"
data["Certificate"]=data["Certificate"].fillna("None")
data["Nudity"]=data["Nudity"].fillna("None")
data["Violence"]=data["Violence"].fillna("None")
data["Profanity"]=data["Profanity"].fillna("None")
data["Alcohol"]=data["Alcohol"].fillna("None")
data["Frightening"]=data["Frightening"].fillna("None")
data.isnull().sum()
```

```python
# box plot on Rate feature
plt.figure(figsize=(20, 6))
sns.boxplot(x = "Rate", data = data)
plt.show()
```

```python
"Find out IQT of Rate feature"
Q1 = data['Rate'].quantile(0.25)
Q3 = data['Rate'].quantile(0.75)
IQR = Q3 - Q1
print("Interquartile range:: ", IQR)
```

```python
Lower = Q1 - 1.5 * IQR
Upper = Q3 + 1.5 * IQR
# remove outliers
data = data[(data['Rate'] > Lower) & (data['Rate'] < Upper)]
data.shape
```

```python
# get boxplot for Votes
plt.figure(figsize=(20, 6))
sns.boxplot(x = "Votes", data = data)
plt.show()
```

```python
# boxplot on Duration

plt.figure(figsize = (20, 6))
sns.boxplot(x = "Duration", data = data)
plt.show()
```

```python
"Remove outliers in Duration"
Q1 = data['Duration'].quantile(0.25)
Q3 = data['Duration'].quantile(0.75)
IQR = Q3 - Q1
print("Interquartile range:: ", IQR)
Lower = Q1 - 1.5 * IQR
Upper = Q3 + 1.5 * IQR
# remove outliers
data = data[(data['Duration'] > Lower) & 
                        (data['Duration'] < Upper)]
data.shape
```

# univariate Analysis

```python
# count plot on Date feature

plt.figure(figsize = (20, 6))
sns.countplot(x = "Date", data = data, 
                order = data["Date"].value_counts().index, 
                            palette = "Set1")
plt.xticks(rotation = 90)
plt.show()
```

```python
plt.figure(figsize = (20, 6))
sns.countplot(x = "Certificate", data = data, 
                order = data["Certificate"].value_counts().index, 
                            palette = "Set2")
plt.xticks(rotation = 90)
plt.show()
```

```python
# distplot on Rate feature
plt.figure(figsize = (20, 6))
sns.distplot(data["Rate"])
plt.show()
```

```python
# distplot on Duration feature
plt.figure(figsize = (20, 6))
sns.distplot(data["Duration"])
plt.show()
```
