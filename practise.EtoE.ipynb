```python
data = pd.get_dummies(data,drop_first = True)
data.head()
```

```python
data.info()
```

```python
data.columns
```

```python
data["sex_male"] = data["sex_male"].replace({True:1,False:0})
data["smoker_yes"] = data["smoker_yes"].replace({True:1,False:0})
data["region_northwest"] = data["region_northwest"].replace({True:1,False:0})
data["region_southeast"] = data["region_southeast"].replace({True:1,False:0})
data["region_southwest"] = data["region_southwest"].replace({True:1,False:0})
data.info()
```

```python
X = data.drop("expenses", axis = 1)
X.head()
```

```python
y = data["expenses"]
y.head()
```

```python
X.describe()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.2, random_state = 25)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
X_train_fit = scaler.fit_transform(X_train)
X_test_fit = scaler.transform(X_test)
```

```python
X_train_fit_df = pd.DataFrame(X_train_fit, columns = X_train.columns)
X_train_fit_df.head()
```

```python
X_train_fit_df.describe()
```

```python
from sklearn.linear_model import LinearRegression 
from sklearn.metrics import r2_score
model_lr = LinearRegression()
model_lr.fit(X_train_fit , y_train)
lr_acc = round(r2_score(y_test , model_lr.predict(X_test_fit)) , 2)
print(f"Linear Regression mofdel accuracy :: {lr_acc*100}")
```

```python
from sklearn.svm import SVR
model_SVR = SVR()
model_SVR.fit(X_train_fit , y_train)
SVR_acc = round(r2_score(y_test , model_SVR.predict(X_test_fit)) , 2)
print(f"SVM  model accuracy :: {SVR_acc*100}")
```

```python
from sklearn.tree import DecisionTreeRegressor
model_dt =DecisionTreeRegressor()
model_dt.fit(X_train_fit , y_train)
dt_acc = round(r2_score(y_test , model_dt.predict(X_test_fit)) , 2)
print(f"Decision Tree Regression mofdel accuracy :: {dt_acc*100}")
```

```python
from sklearn.ensemble import RandomForestRegressor
model_rf =RandomForestRegressor()
model_rf.fit(X_train_fit , y_train)
rf_acc = round(r2_score(y_test , model_rf.predict(X_test_fit)) , 2)
print(f"Random Forest Regression mofdel accuracy :: {rf_acc*100}")
```

```python
models = ["LinearRegression" , "DecisionTreeClassifier" , "Support Vector Machine " , "RandomForestClassifier"]
acc = [ lr_acc , dt_acc ,SVR_acc , rf_acc ]
plt.figure(figsize=(20,10))
sns.barplot(x = acc , y = models ,  palette = "Set2")
Xlabel = "accuracy"
ylabel = "models"
tittle = "....."
plt.show()
```

```python
# save model and scalar parameters
import pickle
pickle.dump(model_rf, open("model_rf_param.pkl", "wb"))
pickle.dump(scaler, open("scaler.pkl", "wb"))
```

```python
# use the saved parameters to predict new data

# Load model
with open("model_RF_param.pkl", "rb") as f:
    model = pickle.load(f)

# Load scaler
with open("scaler.pkl", "rb") as f:
    scaler = pickle.load(f)

# predicr by random forest model on user input
def predict_expense(age, bmi, children, sex, smoker, 
                region_northwest, region_southeast, region_southwest):
    user_input = np.array([[age, bmi, children, sex, smoker, 
                region_northwest, region_southeast, region_southwest]])
    user_input = scaler.transform(user_input)
    return model.predict(user_input)[0]
# new data
# take the user inputs
age = float(input("Enter your age: "))
bmi = float(input("Enter your bmi: "))
children = float(input("Enter your children: "))
sex = float(input("Enter your sex: "))
smoker = float(input("Enter your smoker: "))
region_northwest = float(input("Enter your region_northwest: "))
region_southeast = float(input("Enter your region_southeast: "))
region_southwest = float(input("Enter your region_southwest: "))

# predict the expenses
expenses = predict_expense(age, bmi, children, sex, smoker, 
                region_northwest, region_southeast, region_southwest)
print(f"Predicted expenses: {round(expenses, 2)}")  # round to 2 decimal places if needed {expenses}")
```
