```python
data=data.drop_duplicates()
data.shape
```

```python
data.isnull().sum()
```

```python
#replace rate,vot,duration null with 0 
data['Rate']=data['Rate'].fillna(0)
data['Votes']=data['Votes'].fillna(0)
data['Duration']=data['Duration'].fillna(0)
data.isnull().sum()
```

```python
# in  categorial feactures,replace with none
data['Certificate']=data['Certificate'].fillna('None')
data['Nudity']=data['Nudity'].fillna('None')
data['Violence']=data['Violence'].fillna('None')
data['Profanity']=data['Profanity'].fillna('None')
data['Alcohol']=data['Alcohol'].fillna('None')
data['Frightening']=data['Frightening'].fillna('None')
data.isnull().sum()
```

```python
data.head()
```

```python
select_imdb = data[['Name', 'Rate','Date']]
select_imdb
```

```python
for i in range(0,data.shape[0]):
  data['Votes'][i] = data['Votes'][i].replace(',', '')
data['Votes']
```

```python
data['Rate'] = pd.to_numeric(data['Rate'], errors='coerce', downcast="integer")
data['Votes'] = pd.to_numeric(data['Votes'], errors='coerce', downcast="integer")
data['Duration'] = pd.to_numeric(data['Duration'], errors='coerce', downcast="integer")
```

```python
# box plot on rate fetaure
plt.figure(figsize=(20,6))
sns.boxplot(x ='Rate' , data=data)
plt.show()
```

```python
Q1 = data['Rate'].quantile(0.25)
Q3 = data['Rate'].quantile(0.75)
IQR = Q3 - Q1
print("Interquartile Range ::", IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
data = data[(data['Rate'] >= lower_bound) & (data['Rate'] <= upper_bound)]
data.shape
```
