```python
import pandas as pd
```

```python
col_names = ['sepal_length','sepal_width','petal_length','petal_width','species']
```

```python
csv_url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
flowers_df = pd.read_csv(csv_url, names  = col_names)
flowers_df.head()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot( x='sepal_length', y='sepal_width', data=flowers_df, hue='species', style='species', s=100)
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.scatterplot( x='petal_length', y='petal_width', data=flowers_df)
plt.show()
```
